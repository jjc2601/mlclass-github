# Decision Trees

In this notebook we will take another try at our decision trees, however this time we will be using the Decision Tree model instead of KNN. 

## Setup libraries

```python
# TODO : Import relevant libraries 
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.tree import export_text
import joblib
```

## Step 1: Load our Data

```python
# TODO : Load our data.
data = pd.read_csv("penguins.csv")
```

```python
# TODO : Split the data into testing and training data at a 50/50 split.
y = data["species"]

x = data[["culmen_length_mm","culmen_depth_mm","flipper_length_mm","body_mass_g"]]

X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size = 0.5, random_state=42)
```

## Step 2: Choose your algorithm

You may use [this documentation](https://www.geeksforgeeks.org/machine-learning/building-and-implementing-decision-tree-classifiers-with-scikit-learn-a-comprehensive-guide/) as a reference for how to use the decision tree classifier from sklearn. 

You may skip step 7 for now, today I want you to try to manually adjust the hyperparameters. We will go over how to automate this process next class. 

```python
# TODO select the DecisionTreeClassifier model from sklearn and assign it to a variable

clf = DecisionTreeClassifier(random_state= 1, max_depth= 6)
```

## Step 3: Training

We are finally adding code to our training part of the cycle! Build your tree by "fitting" your decision tree using X_train and y_train. Then you can visualize your decision tree using [export_text](https://scikit-learn.org/stable/modules/generated/sklearn.tree.export_text.html). Alternatively, you can use matplotlib (as shown in the geeksforgeeks examples)


```python
 
 # TODO fit your model to X_train and y_train
clf.fit(X_train, y_train)
# TODO visualize your text with export_text from sklearn or matplotlib
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy}')

r = export_text(clf,feature_names = list(X_train.columns))
print(r)
```

# Step 4: Testing/Tuning

```python
#TODO Paste in your compute_accuracy function
def compute_accuracy(predictions, answers):
    correct = 0 
    for i in range(len(predictions)):
        if predictions[i] == answers[i]:
            correct += 1
    accuracy = round((correct/(len(predictions)) * 100),2)
    return accuracy
```

```python
accuracy = compute_accuracy(y_pred, y_test.tolist())
print(f'Accuracy: {accuracy}')
```

```python
filename = 'decision_tree.pkl'
joblib.dump(clf,filename)
loaded_model = joblib.load(filename)
prediction = loaded_model.predict(x[0:1])
```
